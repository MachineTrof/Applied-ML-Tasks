# Оценка прибыльности

Данные от магазинов
* От 5% до 15% посетителей совершают покупку.
* Средняя стоимость покупки — от 20 до 22 тысяч рублей.
* Наценка на товар — 20%.
* Магазин работает 18 часов в сутки.
* В магазине посменно работают 2 продавца.
* Зарплата продавца — 50 000 рублей + 43% налогов.


## Ожидаемая выручка

$$R = t \times h \times v \times s \times a \times m \times d,$$

где $R$ - ожидаемая выручка в месяц,

$t$ - проходимость,

$h$ - количество рабочих часов в день,

$v$ - доля посетителей от количества прохожих,

$s$ - доля покупателей от количества посетителей,

$a$ - средняя стоимость покупки,

$m$ - наценка,

$d$ - количество рабочих дней в году.


Чтобы оценить потенциальную прибыль, из расчёта выручки сразу убрали себестоимость товаров, использовав только наценку.
Подставим в формулу средние показатели: пусть в магазин заходит каждый 225-й прохожий, каждый 10-й из посетителей совершает покупку, а стоимость покупки составляет 21 000 рублей.
Тогда выручка будет равна:

$$R = t \times 18 \times \frac{1}{225} \times 0.1 \times 21000 \times 0.2 \times 30$$

## Ожидаемые расходы

$$E = p + e \times w,$$

где $E$ - ожидаемые расходы в месяц,

$p$ - стоимость аренды,

$e$ - количество продавцов,

$w$ - зарпата (с налогами)

Подставляем данные и получаем ожидаемые расходы:
$$E = p + 2 \times 50000 \times 1.43$$

## Ожидаемая прибыльность

$$P_r = R - E,$$

где $P_r$ - ожидаемая прибыльность в месяц,

$R$ - ожидаемая выручка в месяц,

$E$ - ожидаемые расходы в месяц

Наконец, объединим две формулы в одну:

$$P_r = R - E = t \times 18 \times \frac{1}{225} \times 0.1 \times 21000 \times 0.2 \times 30 - (p + 2 \times 50000 \times 1.43)$$

Осталось только подставить её в код.

# Расчёт прибыльности

Лучшее помещение для магазина — то, которое принесёт больше всего прибыли. Чтобы отыскать его, нужно рассчитать ожидаемую прибыльность всех 94 подходящих объектов недвижимости из датасета и выбрать самые перспективные.

````python
df.head(1)
````

````python
area = df['area']
price = df['price']
traffic = df['traffic']
address = df['address']
````

1.
Ваш фильтр создаёт четыре списка с ключевыми параметрами помещения. Мы добавили ещё один — `filtered_objects_profits`. Сохраните в нём прогноз прибыли, рассчитанный по формуле:


 $P_r = t \times 18 \times \frac{1}{225} \times 0.1 \times 21000 \times 0.2 \times 30 - (p + 2 \times 50000 \times 1.43),$ где $t$ - значение переменной `traffic`, $p$ - значение переменной `price`.


 Затем выведите максимальное значение из этого списка на экран. Для этого примените функцию `max()`, указав в её скобках название списка — `filtered_objects_profits`.

````python
def Pr(t, p, h=18, v=1/225, s=0.1, a=21000, m=0.2, d=30, e=2, w1=50000, w2=1.43):
    return float(t * h * v * s * a * m * d - (p + e * w1 * w2))
````

````python
filtered_objects_profits = [Pr(traffic[index], price[index]) for index in df.index]
max(filtered_objects_profits)
````

2.
Вы нашли максимальную прибыльность среди подходящих объектов. Теперь определите, какому объявлению соответствует этот показатель.
Для этого пройдите по списку `filtered_objects_profits` и сравните его значения с максимальным. Если они совпадут — выведите на экран индекс объекта.

````python
filtered_objects_profits = [(Pr(traffic[index], price[index]), index) for index in df.index]
max(filtered_objects_profits, key=lambda x: x[0])
````

3.
Одного объекта мало — нужно представить несколько лучших вариантов.
Найдите все объекты, потенциальная прибыльность которых больше 500 тысяч рублей, и выведите на экран всю сохранённую информацию по ним: площадь, стоимость аренды, проходимость, адрес и прогноз прибыли.

````python
for pr, index in filtered_objects_profits:
    if pr > 500000:
        print(area[index], price[index], traffic[index], address[index])
````

4.
Казалось бы, задача решена. Однако формула для прогноза прибыли основана на средних показателях существующих магазинов — а реальные могут оказаться ниже.
Если взять за основу не средние параметры, а минимальные, финальный результат может измениться. Формула точно изменится. Вместо каждого 225-го посетителя — каждый 300-й, вместо каждого 10-го покупателя — каждый 20-й, а вместо среднего чека в 21 000 рублей — 20 000 рублей:

$$P_r = t \times 18 \times \frac{1}{300} \times 0.05 \times 20000 \times 0.2 \times 30 - (p + 2 \times 50000 \times 1.43)$$


Добавим к выводу на экран минимальную ожидаемую прибыль для каждого из пяти лучших объектов.

````python
best = sorted(filtered_objects_profits, key=lambda x: x[0])[::-1][:5]
best
````

````python
best_ind = [i[1] for i in best]
[(Pr(traffic[index], price[index], v=1/300, s=0.05, a=20000), index) for index in best_ind]
````
