# Исследование объявлений о продаже квартир

**Описание**

## Предобработка данных

```python
data['first_day_exposition'] = pd.to_datetime(data['first_day_exposition'])
data['is_apartment'] = data['is_apartment'].astype('boolean')
data['locality_name'] = data['locality_name'].astype('string')
```

```python
list(data['locality_name'].unique())[:10], len(data['locality_name'].unique())
```

```python
def clean_loc(name):
    if pd.isna(name):
        return name
    full_name = name.split()
    capitalized = [w for w in full_name if w and (w[0].isupper() or w[0].isdigit())]
    return " ".join(capitalized)

data['locality_name'] = data['locality_name'].apply(clean_loc)
data['locality_name'] = data['locality_name'].astype('string')
len(data['locality_name'].unique())
```

```python
np.sort(data['ceiling_height'].unique())
```

```python
def clean_ceiling(h):
    if pd.isna(h):
        return h
    if h < 2 or h == 100:
        return np.nan
    if h >= 20:
        return h / 10
    if h > 6:
        return np.nan
    return h

data['ceiling_height'] = data['ceiling_height'].apply(clean_ceiling)
np.sort(data['ceiling_height'].unique())
```

```python
data['balcony'] = data['balcony'].fillna(0).astype('int64')
data['parks_around3000'] = data['parks_around3000'].fillna(0).astype('int64')
data['ponds_around3000'] = data['ponds_around3000'].fillna(0).astype('int64')

data['is_apartment'] = data['is_apartment'].fillna(False)
data['is_apartment'] = data['is_apartment'].astype('bool')
```

```python
kitchen_fraction = (data['kitchen_area'] / data['total_area']).median()
living_fraction = (data['living_area'] / data['total_area']).median()

data.loc[data['kitchen_area'].isna(), 'kitchen_area'] = data['total_area'] * kitchen_fraction
data.loc[data['living_area'].isna(), 'living_area'] = data['total_area'] * living_fraction

kitchen_fraction, living_fraction
```

```python
data.info()
```

## Посчитайте и добавьте в таблицу новые столбцы


```python
data['square_meter_price'] = data['last_price'] / data['total_area']
data['distance_center_km'] = data['cityCenters_nearest'] / 1000

data['weekday_exposition'] = data['first_day_exposition'].dt.weekday
data['month_exposition'] = data['first_day_exposition'].dt.month
data['year_exposition'] = data['first_day_exposition'].dt.year

def floor_type(row):
    f, ft = row['floor'], row['floors_total']
    if f == 1:
        return 'первый'
    if f == ft:
        return 'последний'
    return 'другой'

data['floor_type'] = data[['floor','floors_total']].apply(floor_type, axis=1)

data[['square_meter_price', 'distance_center_km', 'weekday_exposition',
      'month_exposition', 'year_exposition', 'floor_type']].head()
```

```python
data['floor_type'] = data['floor_type'].astype('string')
data['weekday_exposition'] = data['weekday_exposition'].astype('int64')
data['month_exposition'] = data['month_exposition'].astype('int64')
data['year_exposition'] = data['year_exposition'].astype('int64')
data.info()
```

## Проведите исследовательский анализ данных

```python
q1, q3 = data['days_exposition'].dropna().quantile([0.25, 0.75])
q1, q3
```

```python
data[['last_price','total_area','living_area','rooms']].corr()['last_price']
```

```python
top10 = data['locality_name'].value_counts().head(10).index
avg_price = (data[data['locality_name'].isin(top10)]
             .groupby('locality_name')['square_meter_price']
             .mean().sort_values(ascending=False))
avg_price
```
