# Оценка прибыльности

Данные от магазинов
* От 5% до 15% посетителей совершают покупку.
* Средняя стоимость покупки — от 20 до 22 тысяч рублей.
* Наценка на товар — 20%.
* Магазин работает 18 часов в сутки.
* В магазине посменно работают 2 продавца.
* Зарплата продавца — 50 000 рублей + 43% налогов.


## Ожидаемая выручка

$$R = t \times h \times v \times s \times a \times m \times d,$$

где $R$ - ожидаемая выручка в месяц,

$t$ - проходимость,

$h$ - количество рабочих часов в день,

$v$ - доля посетителей от количества прохожих,

$s$ - доля покупателей от количества посетителей,

$a$ - средняя стоимость покупки,

$m$ - наценка,

$d$ - количество рабочих дней в году.


Чтобы оценить потенциальную прибыль, из расчёта выручки сразу убрали себестоимость товаров, использовав только наценку.
Подставим в формулу средние показатели: пусть в магазин заходит каждый 225-й прохожий, каждый 10-й из посетителей совершает покупку, а стоимость покупки составляет 21 000 рублей.
Тогда выручка будет равна:

$$R = t \times 18 \times \frac{1}{225} \times 0.1 \times 21000 \times 0.2 \times 30$$

## Ожидаемые расходы

$$E = p + e \times w,$$

где $E$ - ожидаемые расходы в месяц,

$p$ - стоимость аренды,

$e$ - количество продавцов,

$w$ - зарпата (с налогами)

Подставляем данные и получаем ожидаемые расходы:
$$E = p + 2 \times 50000 \times 1.43$$

## Ожидаемая прибыльность

$$P_r = R - E,$$

где $P_r$ - ожидаемая прибыльность в месяц,

$R$ - ожидаемая выручка в месяц,

$E$ - ожидаемые расходы в месяц

Наконец, объединим две формулы в одну:

$$P_r = R - E = t \times 18 \times \frac{1}{225} \times 0.1 \times 21000 \times 0.2 \times 30 - (p + 2 \times 50000 \times 1.43)$$

Осталось только подставить её в код.
